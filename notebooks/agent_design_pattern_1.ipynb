# Agentic AI - Architecture

Agentic AI can learn and adapt, not just follow a rigid script. It detects, understands and adapts with the change in the environment. The Agentic AI architecture, is a structural arrangement of an agentic system that defines the interactions among several elements to accomplish intelligent activity.

- Enables complex behavior with stuctural breakdown. (Simplification for Easy Maintenance)

- Facilitates Learning and Apaptation. (Scalability)

- Enhances Robustness. (Security, Performance and Reliability)

- Promotes Reusability. (Low Maintanance)

## 3 Main Components of Agentic System

- **Prompt**: Outlines the agent's functionining principles, comprising particular objectives and limitations.

- **Memory**: Serves as the agent's store of knowledge, enablisng learn from previous interactions, and make well-informed choices.

- **Tools**: These are outside resources, like APIs or functions that allows the agent to execute different tasks and active its goals.


## Agentic AI Architecture Layers

### Core Layers


```
Environment
   ↓
Perception (state)
   ↓
Cognition (reason + decide)
   ↓
Action (execute)
   ↓
Feedback → Perception
```

#### 1. Perception Layer

- Observes the environment and turns raw signals (user text, logs, metrics, events or API responses) into a clean, structured understanding of “what is happening right now.”

- It senses, interprets, and structures the signal into a structured message. It does **NOT decide** what to do.

- Why It's needed?
    - Logs are noisy
    - User inputs are ambiguous
    - Events arrive out of order
    - Metrics are continious numbers
    - data may need transformation

- Perception Layer filters noise, extracts meaning and buils a state snapshot.

**Example** (DevOps / Azure Agent)
Input: Raw Inputs (App Insight logs, Azure Monitoring metrics, and Deployment Events)

Perception Layer: Correlates spike with deployment timestamp, detects regression patters, and classifies severity

Output:
```json
{
  "incident": true,
  "affected_service": "api-gateway",
  "trigger": "deployment",
  "severity": "P1",
  "recommended_focus": "rollback_or_scale"
}

```

#### 2. Cognitive Layer

- It is the **Thinking System**. It decides what the input state means, and what to do.

- Planning, Agent Orchestration, Memory Management, Continious Learing, Decission Making, etc...

- Cognitive Layer interprets the perceived state, reasons about it, selects goals, plans actions, and decides what to do next under constraints and uncertainty.

- Reasoning systles could be chain-of-thoughts, ReAct, Reflection loops...

Input (Output from Perception Layer):
```json
{
  "incident": true,
  "service": "api-gateway",
  "severity": "P1",
  "deployment_recent": true
}

```

Cognitive Reasoning:
- Recent deployment - likely caused it
- P1 severity - act fast
- Rollback is reversible

Output:
```json
{
  "action": "rollback",
  "approval_required": true,
  "confidence": 0.92
}

```

#### 3. Action Layer

- It is the **Execution System**.

- The Action Layer safely turns decisions from the cognitive layer into real-world effects/actions and reports what actually happened.

Input (output from Cognitive Layer):
```json
{
  "goal": "restore_service",
  "strategy": "rollback",
  "target": "api-gateway",
  "approval_required": true
}

```

Output (take action and confirm):
```json
{
  "action": "rollback",
  "status": "success",
  "duration_seconds": 41,
  "side_effects": ["cache_flushed"],
  "confidence": 0.99
}

```

#### 4. Learning Layer (optional)

```
Perception → Cognition → Action
     ↑            ↓
     └── Learning ←─ Outcome
```

- It is the **Adaptation Mechanism** of the agent.

- The Learning Layer observes outcomes, evaluates decisions, and updates the agent’s internal knowledge, policies, or models so future decisions improve.

- Without it, Agents repeat mistakes and Behavior is static.

- Memory management or Fine tunning or RAG plays a role here.

- It increases accuracy, efficiency and judgement.

#### 5. Collaborative Layer (incase of Multi-agent)

- Enables agent ability to solve the problem alone or in collaboration with other agents or Human-agents.

- Allows agents to share messages and informations. 

- Supports parallel as well as sequential cooperation across settings.

#### 6. Security Layer

- The Security Layer ensures the agent can only see, decide, and act within explicitly allowed boundaries—and nothing more.

- User authenticatio and authorization

- Prevents dangerous execution using Guardrails.

- Data Security and Auditing

### Standard Layers

#### 1. Integration Layer

- Connects Agentic AI layers and external systems for communication and data flow.

- Enterprise APIs, Human-in-loop interface, LLM Hubs and Database & Knowledgebase.

#### 2. Operations Layer

- manages realtime performance, monitors activities, and enables continuous improvements.

- Performance & Evaluation, Resource & Tokenization, Agent Trace & tracking, Guardrails for Responsive AI, Monitoring & Alerts.

#### 3. Infrastructure Layer

- Provides computational power and storage for efficient and reliable agent execution.

- Runtime, Vector Database, LLM APIs, Data Engineering Pipelines.

# Agentic AI - Design Patterns

## Reflection Pattern

The **Reflection Pattern** enables an agent to step back, evaluate its own thinking or actions, and improve before proceeding—adding meta-cognition to agentic AI systems.

It helps by:
- Reducing hallucination

- Improving reasoning quality by self-examination.

- Catching logical inconsitency

- Increasing relability in complex task

Reflection loop: `Plan → Act → Observe → Reflect → Revise → Act`

## Tool Use Patterns

In agentic AI, **Tool Use Patterns** describe _how an agent decides to invoke external tools (APIs, functions, databases, services), uses their outputs, and integrates results back into reasoning and actions._

**1. Direct Tool Invovation Pattern**
- The agent calls a tool once to fetch or compute something, then responds. `User → Agent → Tool → Agent → Final Answer`

- Example, A weather forecasting agent invoking Weather API. It MUST Always call the API to forecast.

**2. ReAct (Reason + Act Interleaving)**

- The agent reasons first, then decides whether and which tool to use. `Reason → Decide Tool → Call Tool → Answer`

- Example, An agent has 2 tools to its disposal. (one reads RAG dB, one searches Web). Agents decides on user query which tool to call and makes the call. If the query about company policy, it reads RAG. If the query about latest news, it goes to Web.

**3. Parallel Tool Use (Fan-Out / Fan-In)**

- Agent calls multiple tool in parallel.

**4. Human-in-the-Loop Tool Use**

- Agent pauses before tool execution for approval.

**5. Tool-Chaining (Pipeline Pattern)**

- Output of one tool feeds into another. `Tool A → Tool B → Tool C`
