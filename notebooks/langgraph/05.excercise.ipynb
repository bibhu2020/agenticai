```python
from typing import List, TypedDict, Sequence
from langgraph.graph import StateGraph
```

```python
# Define the state schema
class AgentState(TypedDict):
    values: Sequence[int]
    name: str
    result: str

# Define the tool function
def process(state: AgentState) -> AgentState:
    """
    Process the values in the state and return a new state with the result.
    """
    print(state )
    # Calculate the sum of values
    total = sum(state['values'])
    
    # Update the result in the state
    state['result'] = f"Hi there {state['name']}, the sum of your values is {total}."
    print(state )
    return state    

# define the graph
graph = StateGraph(AgentState)

graph.add_node("processor", process)
graph.set_entry_point("processor")
graph.set_finish_point("processor")

app = graph.compile()
```

```python
# Display the graph in Jupyter Notebook
from IPython.display import Image, display
display(Image(app.get_graph().draw_mermaid_png()))
```

```python
result = app.invoke(AgentState(name="John", values=[1, 2, 3, 4, 5]))
print(result['result'])  # Output: John, you are doing amazazing job learning LangGraph!
```
